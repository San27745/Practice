```python
# Two Threads
from threading import Thread
import time 
def numbers():
    for i in range(1, 6):
        print(i)
        time.sleep(1)
def alphabets():
    for ch in ['A', 'B', 'C', 'D', 'E']:
        print(ch)
        time.sleep(2)
t1 = Thread(target=numbers)
t2 = Thread(target=alphabets)
t1.start()
t2.start()
t1.join()
t2.join()
print("Finished")
```

```python
from threading import Thread
from time import sleep
def even():
    for i in range(2, 11, 2):
        print("Even:", i)
        sleep(1)
def odd():
    for i in range(1, 10, 2):
        print("Odd:", i)
        sleep(1)
t1 = Thread(target=even)
t2 = Thread(target=odd)
t1.start()
t2.start()
t1.join()
t2.join()
print("Number Generation Completed")
```

```python
from threading import Thread
import time
def square(num):
    print(f"Square: {num * num}")
    time.sleep(1)
def cube(num):
    print(f"Cube: {num * num * num}")
    time.sleep(2)
# Create threads using threading module.thread function 
t1 = Thread(target=square, args=(4,)) #args takes tuple so give ,comma to pass 1 argument
t2 = Thread(target=cube, args=(4,))

t1.start()
t2.start()

t1.join()
t2.join()
print("Done!")
```

```python
'''Create three threads to perform:
Product Search
Add to Cart
Payment Processing
Use sleep() to simulate processing time and join() to wait for all tasks.'''
import threading
import time
# Function for Product Search
def product_search():
    print("Searching products...")
    time.sleep(2)
    print("Products found successfully.")
# Function for Add to Cart
def add_to_cart():
    print("Adding product to cart...")
    time.sleep(3)
    print("Product added to cart successfully.")
# Function for Payment Processing
def payment_processing():
    print("Processing payment...")
    time.sleep(4)
    print("Payment completed successfully.")
print("Online Shopping Started...\n")
# Create Threads
t1 = threading.Thread(target=product_search)
t2 = threading.Thread(target=add_to_cart)
t3 = threading.Thread(target=payment_processing)
# Start Threads
t1.start()
t2.start()
t3.start()
# Wait for all threads to complete
t1.join()
t2.join()
t3.join()
print("\nOrder placed successfully.")
print("Thank you for shopping!")
```

```python
import threading
import time
 
orders_processed = 0
lock = threading.Lock()
 
# Order Verification
def verify_order():
    global orders_processed
 
    for i in range(3):
        print(f"{threading.current_thread().name}: Verifying Order {i+1}")
        time.sleep(2)
 
        lock.acquire()
        orders_processed += 1
        lock.release()
 
# Payment Processing
def process_payment():
    global orders_processed
 
    for i in range(3):
        print(f"{threading.current_thread().name}: Processing Payment {i+1}")
        time.sleep(1)
 
        lock.acquire()
        orders_processed += 1
        lock.release()
 
# Inventory Update
def update_inventory():
    global orders_processed
 
    for i in range(3):
        print(f"{threading.current_thread().name}: Updating Stock {i+1}")
        time.sleep(1.5)
 
        lock.acquire()
        orders_processed += 1
        lock.release()
 
# Creating Threads
t1 = threading.Thread(target=verify_order,
                      name="Verification Thread")
 
t2 = threading.Thread(target=process_payment,
                      name="Payment Thread")
 
t3 = threading.Thread(target=update_inventory,
                      name="Inventory Thread")
 
# Starting Threads
t1.start()
t2.start()
t3.start()
 
# Waiting for Completion
t1.join()
t2.join()
t3.join()
print("\nAll threads completed.")
print("Total Operations Processed:", orders_processed)
```
