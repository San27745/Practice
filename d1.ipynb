```python
arr0=numpy.array(78)
print(arr0)
print(arr0.ndim)
```

ndim is used to give dimension if there is [11,22,33,44] one dimension and 2 dimension has [[11,22,33],[11,22,33]] 3 dimension has [[[11,22],[11,33]]]
shape is used to display row and columns (rows,columns) format
size display number of elements in it

```python
ar1=numpy.array([11,22,33,44])
ar2=numpy.array([[11,22,33],[11,22,33]])
ar3=numpy.array([[[11,22],[11,33],[11,22]]])
print(ar1.ndim)
print(ar2.ndim)
print(ar3.ndim)
print(ar2.shape)
print(ar2.size)
```
