```python
a2=np.ones((3,2))
print(a2)
a3=np.zeros((2,2))
print(a3)
a4=np.full((3,3),4)
print(a4)
a5=np.arange(1,6,2)
print(a5)
a6=np.eye(3,2)
print(a6)
a7=np.linspace(1,10,4)
print(a7)
```

ones print (rows,columns) all element in one format
and zeros print (rows,columns) all element in zero format
full prints ((rows,columns),number)  instead of zero  or one whatever number we pass it prints
arrange prints in range like(startvalue,stop,gap)stop value is stop before it means if we pass 6 it print only 5
eye  we have to pass(rows,columns) if we only pass 3 it prints index matrix
linspace will give range and give that divide in how many parts of it
means (1,10,5) means 1 to 10 in 5 parts
flattern combine 2dimensional araay in one 
std gives standard deviation of array

```python
a1=np.array([1,7,6,3,9,2])
a2=np.sort(a1)
print(a1)   #unsorted
print(a2)   #sorted
```

```python
a1=np.array([1,7,6,3,9,2])
a1.sort()
print(a1)   #ascending sorting
print(a1[::-1]) #descending sorting
```

```python
a1=np.array([1,7,6,3,2])
np.argsort(a1)  #instead of giving sorted array it gives index number  array([0, 4, 3, 2, 1])
```
